### Sending Prompts and Receiving JSON Classifications

We will prepare the necessary prompts to send to the OpenAI GPT-4o-mini model. Each prompt will include the abstract and any other relevant context needed for classification and receive a JSON response. The multiple JSON files containing classified abstracts will be saved into .csv and further analyzed in subsequent steps. This modular approach allows for more robust data processing and minimizes the risk of data loss due to connection issues.


````python
# JSON example structure as a separate variable
json_example = '''{
  "classification": "Category",
  "keywords": [],
  "rationale": "Detailed explanation linking abstract and examples",
  "Feeling AI":"X%",
  "Mechanical AI":"X%",
  "Thinking AI":"X%",
  "Out of Scope":"X%",
  "Not Identified":"X%"
}'''

# Creation of the FewShotPromptTemplate
prompt = PromptTemplate.from_template(prompt_template)

openai_api_key = '<open-ai key>'  # Set your OpenAI API key
modelo = 'gpt-4o-mini'

# OpenAI model definition
llm_openai = ChatOpenAI(
    temperature=0.7,
    openai_api_key=openai_api_key, 
    model=modelo,
    top_p=1.0,
    frequency_penalty=0.0,
    presence_penalty=0.0
)

llm_chain = prompt | llm_openai

iteration_count = 0

# Modified function to process abstracts
def query_llm(abstract, json_example):  # Removed json_example parameter
    response = llm_chain.invoke({
        "Abstract": abstract,
        "json_example": json_example  # Variável adicionada
    })    
    
    # Improved JSON extraction
    try:
        json_str = response.content.split('```json')[1].split('```')[0].strip()
    except IndexError:
        json_str = response.content.strip()
    
    # Convert to dictionary
    response_dict = json.loads(json_str)
    return response_dict  # Return actual object instead of string

def apply_llm(row):    
    abstract = row['Abstract']    
    response_dict = query_llm(abstract, json_example)  # Removed json_example argument
    global iteration_count
    iteration_count += 1
    
    # Safe column assignment
    json_example_dict = json.loads(json_example)
    for key in json_example_dict.keys():
        row[key] = response_dict.get(key, None)
    
    print(f"Iteration: {iteration_count}")
    return row

def save_to_csv(df, iteration):
    filename = f'df_step_{iteration}.csv'
    df.to_csv('data/articles/openai/'+filename, index=False)
    print(f"Saved {filename}")

# Process the DataFrame in chunks
# chunk_size = 2
# num_chunks = 2

# for i in range(num_chunks):
#     start_idx = i * chunk_size
#     end_idx = min((i + 1) * chunk_size, len(df_all))
#     df_step = df_all.iloc[start_idx:end_idx].copy()
#     df_step = df_step.apply(apply_llm, axis=1)
#     save_to_csv(df_step, i)

# print('End')
# Process the DataFrame in chunks of 200 rows
chunk_size = 100
num_chunks = (len(df_all) // chunk_size) + 1

for i in range(num_chunks):
    start_idx = i * chunk_size
    end_idx = min((i + 1) * chunk_size, len(df_all))
    df_step = df_all.iloc[start_idx:end_idx].copy()
    df_step = df_step.apply(apply_llm, axis=1)
    save_to_csv(df_step, i)
print('End')
````
